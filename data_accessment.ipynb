```python
# We can find the companies info in the corporate_client table, hence we will work with that
# Opening the corporate client table

corporate_clients
```

```python
# Create DF to hold companies
# companies = pd.DataFrame(unique_companies, columns=['Name'])
# Recreating the companies table. Realized the industry type column is in the branch table creating dublication of the industry type
# and that info there is redundant
# In view of that I am moving the industry column to the companies table. Get rid of the DF and fetch the columns from the corporate_clients table.
# Hence no need to get unique names. just drop duplicates

companies = corporate_clients[['Company_Name', 'Industry']].drop_duplicates()
companies 
```

```python
# In SQL, joins are ideally done on ids not on names.  so we will be generating unique ids for them.
# add ID column before the name column
# for the code below, 
# companies.insert - means I am inserting a new coloumn to the companies table
# (0, ID) - means i want to start inserting IDs from the 0 index which is the first record because python starts indexing from 0
# ['CORP' + str(i).zfill(3)] - going throught the all the tables, i noticed most ids start with 'CORP', so i want to do same for the companies table.
# the range starts from 1 to the length of the records in the table companies + 1

companies.insert(0, 'ID', ['CORP' + str(i).zfill(3) for i in range(1, len(companies)+1)])
```

Create Branch Table
taking a careful look at the coperate client table, it contains both information about companies branches and also the policy table. want to seperate them so joining can be easier and optimized.

```python
# Creating the new branch table from the corporate clients. As i do so, i need to also make sure that i include a cloumn that can be used to join the companies table i have created
# To do this, i will add the column company_name since its the only common column for branch and companies 
# Selecting need columns

branch_new = corporate_clients[['Branch_ID', 'Company_Name', 'Region', 'Number_of_Staff']] 
branch_new
```

```python
# But in SQL it is ideal to join tables using IDs instead of names and so i would add the companies id column to the branch table. 
# to do this, i have to first merge the tables branches and companies to create a new branch table. 
# There i will now have the id column and then delete the company name from the table
# from the number of rows(70 rows), it tells the merge was successful

new_branch_t = branch_new.merge(companies, left_on='Company_Name', right_on='Company_Name', how='inner')
new_branch_t  
```

```python
# create function to load tables
# write_mode="WRITE_APPEND" is best used for building an ETL pipeline that loads only new records.
# but for a portfolio project in the development stage or when i am not sure all the records will load correectly at a goal,
# write_mode="WRITE_TRUNCATE" is usually the safer default
# Refer to note in next markdown cell for more info...

def load_dfs_to_bigquery(tables, project_id, dataset_id, write_mode="WRITE_TRUNCATE"):
    client = bigquery.Client(project=project_id)

    job_config = bigquery.LoadJobConfig(
        write_disposition = write_mode,
        autodetect=True
    )

    for table_name, df in tables.items():
        table_id = f"{project_id}.{dataset_id}.{table_name}"

        job = client.load_table_from_dataframe(
            df,
            table_id,
            job_config=job_config
        )

        job.result()

        print(f"Loaded {len(df)} rows into {table_id}")
```
